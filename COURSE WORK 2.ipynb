````python
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.metrics import ConfusionMatrixDisplay


   

DATASET_DIR = Path('/content/drive/MyDrive/cw2/images')  #change youre dataset path here 

SEED =                                                   # fill the seed 
np.random.seed(SEED)

TRAIN_SIZE = 0.
TEST_SIZE = 0.

face_app = FaceRecognizerWrapper(
    img_size=300,
    n_components=50,
    seed=SEED
)

face_app.load_and_preprocess_data(DATASET_DIR)
#  SPLIT

if not np.isclose(TRAIN_SIZE + TEST_SIZE, 1.0):
    raise ValueError("TRAIN_SIZE + TEST_SIZE must equal 1.0")

(
    face_app.X_train,
    face_app.X_test,
    face_app.y_train,
    face_app.y_test
) = train_test_split(
    face_app.X,
    face_app.y,
    train_size=TRAIN_SIZE,
    test_size=TEST_SIZE,
    random_state=SEED,
    stratify=face_app.y
)

print(f"\nTraining proportion: {TRAIN_SIZE * 100:.0f}%")
print(f"Testing proportion: {TEST_SIZE * 100:.0f}%")
print("Training images:", len(face_app.X_train))
print("Testing images:", len(face_app.X_test))
````

````python
# task 2.2 (PCA / SVD)
#  fill in the missing values.


N_COMPONENTS = 0             #  number of principal components
N_EIGENFACES_TO_SHOW = 0     #  number of eigenfaces to visualize


face_app.fit_pca(n_components=N_COMPONENTS)

face_app.plot_eigenfaces(
    n_components=N_EIGENFACES_TO_SHOW
)

# inspect the resulting feature dimensions.
print("\nPCA completed.")
print("Training features before PCA:", face_app.X_train.shape)
print("Testing features before PCA:", face_app.X_test.shape)
print("Training features after PCA:", face_app.X_train_pca.shape)
print("Testing features after PCA:", face_app.X_test_pca.shape)
````

````python

# task 2.3 multi-class classification
face_app.train_classifier()

accuracy, report, conf_matrix, y_pred = face_app.evaluate()

print(f"\nTest Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")
print(report)

fig, ax = plt.subplots(figsize=(7, 6))

ConfusionMatrixDisplay(
    confusion_matrix=conf_matrix,
    display_labels=face_app.class_names
).plot(
    ax=ax,
    cmap=plt.cm.Blues,
    values_format='d'
)

plt.title("Confusion Matrix - Eigenfaces + Logistic Regression")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
